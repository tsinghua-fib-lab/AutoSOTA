```python
import anndata as ad
import pandas as pd
import numpy as np

import umap
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import calinski_harabasz_score

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
input_dir = "/Users/apple/Desktop/KB/data"
adata_train = ad.read_h5ad(input_dir+'/BiddyData/Biddy_train.h5ad')
adata_test = ad.read_h5ad(input_dir+'/BiddyData/Biddy_test.h5ad')

train_labels = adata_train.obs["clone_id"].to_numpy()
test_labels = adata_test.obs["clone_id"].to_numpy()


# train_embeddings = np.load(input_dir+'/feat_RECOMB/train_test/biddy_train_test/Biddy_scvi_train_embeddings.npy')
# test_embeddings = np.load(input_dir+'/feat_RECOMB/train_test/biddy_train_test/Biddy_scvi_test_embeddings.npy')

# print(train_labels.shape, test_labels.shape)
# print(train_embeddings.shape, test_embeddings.shape)
```

### Random guess baseline

```python
# ---------------------------
# Random guess baseline
# ---------------------------
np.random.seed(42)

# Sample labels according to the training lineage distribution
values, counts = np.unique(train_labels, return_counts=True)
probs = counts / counts.sum()

# Random predictions for test cells
y_pred = np.random.choice(values, size=len(test_labels), p=probs)

# Accuracy and error
accuracy = accuracy_score(test_labels, y_pred)
error = 1 - accuracy

print(f"Random-guess test accuracy: {accuracy * 100:.2f}%")
print(f"Random-guess test error: {error:.3f}")
```
