```python
import pandas as pd
import ast
import matplotlib.pyplot as plt

# List of CSV files and corresponding model names
csv_files = [
    "coverage_deepseek-llm-67b-base.csv",
    "coverage_Llama-2-70b-hf.csv",
    "coverage_Qwen-72B.csv",
    "coverage_Yi-34B.csv"
]
model_names = [
    "Deepseek-LLM-67B",
    "Llama-2-70B-HF",
    "Qwen-72B",
    "Yi-34B"
]

# Function to extract values from the string-encoded dict columns
def extract_val(series, col):
    return series[col].apply(ast.literal_eval).apply(lambda d: next(iter(d.values())))

# --- Plot 1: Coverage Difference (coverage - 0.9) ---

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharey=True)

for ax, csv_path, name in zip(axes.flatten(), csv_files, model_names):
    df = pd.read_csv(csv_path)
    df['coverage_LAC_val']   = extract_val(df, 'coverage_LAC')
    df['coverage_LAC_W_val'] = extract_val(df, 'coverage_LAC_W')
    
    ax.violinplot(
        [df['coverage_LAC_val'] - 0.9, df['coverage_LAC_W_val'] - 0.9],
        showmeans=True
    )
    ax.set_xticks([1, 2])
    ax.set_xticklabels(['LAC', 'LAC_W'])
    ax.set_title(name)
    ax.axhline(0, color='black', linestyle='--', linewidth=0.8)
    ax.set_ylabel('Coverage − 0.9')

fig.suptitle('Coverage Difference Across Models')
plt.tight_layout()
plt.show()


# --- Plot 2: Set Size Comparison ---

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharey=True)

for ax, csv_path, name in zip(axes.flatten(), csv_files, model_names):
    df = pd.read_csv(csv_path)
    df['setsize_LAC_val']   = extract_val(df, 'setsize_LAC')
    df['setsize_LAC_W_val'] = extract_val(df, 'setsize_LAC_W')
    
    ax.violinplot(
        [df['setsize_LAC_val'], df['setsize_LAC_W_val']],
        showmeans=True
    )
    ax.set_xticks([1, 2])
    ax.set_xticklabels(['LAC', 'LAC_W'])
    ax.set_title(name)
    ax.set_ylabel('Set Size')

fig.suptitle('Set Size Comparison Across Models')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import ast
import matplotlib.pyplot as plt

# List of CSV files and corresponding model names
csv_files = [
    "coverage_deepseek-llm-67b-base.csv",
    "coverage_Llama-2-70b-hf.csv",
    "coverage_Qwen-72B.csv",
    "coverage_Yi-34B.csv"
]
model_names = [
    "Deepseek-LLM-67B",
    "Llama-2-70B-HF",
    "Qwen-72B",
    "Yi-34B"
]

# Helper to extract numeric coverage
def extract_cov(series, col):
    return series[col].apply(ast.literal_eval).apply(lambda d: next(iter(d.values())))

# Create 2×2 paired scatter-plots
fig, axes = plt.subplots(2, 2, figsize=(10, 10), sharex=True, sharey=True)

for ax, path, name in zip(axes.flatten(), csv_files, model_names):
    df = pd.read_csv(path)
    df['d_LAC']   = extract_cov(df, 'coverage_LAC')   - 0.9
    df['d_LAC_W'] = extract_cov(df, 'coverage_LAC_W') - 0.9
    
    mask = df['d_LAC'] < 0
    
    ax.scatter(df['d_LAC'], df['d_LAC_W'], alpha=0.3, label='All points')
    ax.scatter(df.loc[mask, 'd_LAC'], df.loc[mask, 'd_LAC_W'], alpha=0.6, label='LAC < 0')
    
    lims = [
        min(df[['d_LAC', 'd_LAC_W']].min().min(), 0),
        max(df[['d_LAC', 'd_LAC_W']].max().max(), 0)
    ]
    ax.plot(lims, lims, linestyle='--', linewidth=1)
    
    ax.set_title(name)
    ax.set_xlabel('LAC: coverage − 0.9')
    ax.set_ylabel('LAC_W: coverage − 0.9')
    ax.legend(loc='upper left')

fig.suptitle('Paired Coverage Differences (LAC vs. LAC_W)')
plt.tight_layout()
plt.show()
```
