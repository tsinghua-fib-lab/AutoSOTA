```python
# 2025.12.22
# Correlation of growth rate
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('data/simulation_gene_data.csv')
X = np.array(data)[:,1:]
X1 = X[:,0]
X2 = X[:,1]
g_true =  X2**2/(1+X2**2)
plt.scatter(X1,X2,c=g_true)
```

```python
from scipy.stats import spearmanr, pearsonr
for delta in ['_1.3','_1.5','_2','_2.5','_3']:
    rs = []
    for seed in [113,4135,0,42,1774952,7266,3682,76458,11451,798]:
        g_predict = np.load(f'results/k_value_{seed}{delta}.npy')[:,0]
        rs.append(pearsonr(g_true, g_predict)[0])

    print(np.mean(rs),np.std(rs))
```
