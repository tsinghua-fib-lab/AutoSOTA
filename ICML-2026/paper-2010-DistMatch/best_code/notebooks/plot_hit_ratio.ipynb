```python
from matplotlib import pyplot as plt
```

```python
top_k_data = {
    "Elec": {"topk": {3: 0.00, 5: 0.00, 7: 0.00,}, "left": 1.00},
    "Solar": {"topk": {3: 0.70, 5: 1.00, 7: 1.00,}, "left": 0.52},
    "Wind": {"topk": {3: 1.00, 5: 0.00, 7: 0.00,}, "left": 1.00},
    "META": {"topk": {3: 0.00, 5: 0.00, 7: 1.00,}, "left": 1.00},
    "NVDA": {"topk": {3: 1.00, 5: 1.00, 7: 1.00,}, "left": 1.00},
    "rain": {"topk": {3: 0.00, 5: 0.90, 7: 1.00,}, "left": 1.00},
}
```

```python
fig, axes = plt.subplots(1, len(top_k_data), figsize=(20, 3))

for idx, (dataset, stats) in enumerate(top_k_data.items()):
    ax = axes[idx]
    top_k_stats = stats["topk"]
    xs = top_k_stats.keys()
    ys = top_k_stats.values()
    x_min, x_max = min(xs), max(xs)

    ax.plot(xs, ys, linewidth=2, marker='o', label="Top-K hit ratio")
    ax.hlines(stats["left"], x_min, x_max, linestyles=['dashed'], colors='red', label="Leftmost hit ratio")
    ax.set_ylim(-0.05, 1.05)
    ax.set_title(dataset)
    if idx == 0:
        ax.set_ylabel('Hit ratio')
    ax.set_xlabel('K')
    ax.legend()
```
