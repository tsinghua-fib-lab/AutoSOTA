```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point


def get_embedding_for_latlon(lat, lon, features_df, zip_polygons_gdf):
    # Step 1: Create a Point object for the input coordinates
    point = Point(lon, lat)  # (lon, lat) order for shapely

    # Step 2: Find the ZIP polygon containing the point
    match = zip_polygons_gdf[zip_polygons_gdf.contains(point)]

    if match.empty:
        return None  # No ZIP found for this coordinate

    # Step 3: Get the ZIP code (e.g., "zip/63621")
    zip_code = match.iloc[0]["place"]

    # Step 4: Lookup that ZIP in the features dataframe
    row = features_df[features_df["place"] == zip_code]

    if row.empty:
        return None  # ZIP exists in geo but not in features data

    # Step 5: Extract only the feature columns
    feature_columns = [col for col in features_df.columns if col.startswith("feature")]
    embedding = row[feature_columns].values[0]  # Get the numpy array

    return embedding


embedding = get_embedding_for_latlon(
    lat=41.102934, lon=-119.695361, features_df=zcta, zip_polygons_gdf=gdf_county
)
```
