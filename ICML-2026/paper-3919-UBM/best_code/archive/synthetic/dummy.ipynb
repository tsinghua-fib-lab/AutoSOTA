```python
import numpy as np
import pandas as pd

def sample_ps(n_samples=10000):
    n_samples = 1000
    n_samples_per_interval = n_samples // 2

    samples_interval1 = np.random.uniform(0, 1, n_samples_per_interval)
    samples_interval2 = np.random.uniform(0, 1, n_samples_per_interval)
    samples = np.concatenate([samples_interval1, samples_interval2])

    return np.random.permutation(samples)


def sample_vars(n_samples):
    y0 = sample_ps(n_samples) 
    y1 = sample_ps(n_samples) 

    s0 = sample_ps(n_samples) 
    s1 = sample_ps(n_samples) 

    a0 = sample_ps(n_samples) 
    a1 = sample_ps(n_samples) 

    df = pd.DataFrame({"Y0": y0, "Y1": y1, "S0": s0, "S1": s1, "A0": a0, "A1": a1})

    df['VarA'] = df.apply(lambda r: ((r["A0"] + r["A1"]) / 2) * (1 - (r["A0"] + r["A1"]) / 2), axis=1)
    df['VarS'] = df.apply(lambda r: ((r["S0"] + r["S1"]) / 2) * (1 - (r["S0"] + r["S1"]) / 2), axis=1)

    df['Bias'] = df.apply(lambda r: abs(r["Y1"] - r["Y0"]) * abs(r["A1"] - r["A0"]) / (2 * (r["A1"] + r["A0"])), axis=1)

    return df
```

```python
n_samples = 1000000
df = sample_vars(n_samples)
```
