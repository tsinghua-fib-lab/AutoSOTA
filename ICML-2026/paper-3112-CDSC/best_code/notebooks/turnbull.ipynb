# Sample Code to Run Turnbull Estimator

## Prepare Data

```python
import numpy as np

lb = np.array([0, 2, 0, 1, 3, 1, 1, 1, 2, 2])
ub = np.array([2, np.inf, 1, 3, np.inf, 2, 2, 2, 3, 3])
```

## Run Turnbull Estimator

```python
import cenreg.model.nonparametric

dist = cenreg.model.nonparametric.turnbull_estimator(lb, ub)
```

## Plot Estimated Distribution

```python
%matplotlib inline

import matplotlib.pyplot as plt

x = dist.b
w = np.diff(x)
cum_p = np.append(0.0, dist.cum_p)
p = np.diff(cum_p)

fig = plt.figure()
plt.xlabel('y')
plt.ylabel('probability')
plt.xlim([dist.b[0], dist.b[-1]])
plt.ylim([0.0,1.0])
plt.bar(x[:-1], p, width=w, align='edge')
plt.title('Probability mass function of the Turnbull estimator')
plt.show()
```
