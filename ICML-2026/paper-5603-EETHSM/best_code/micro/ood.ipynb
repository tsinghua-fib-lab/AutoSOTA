```python
# List of commands to run
commands = []

def make_command(layer1, layer2, embed_dim, window, num_heads, state_dim, run_number=0):
    cmd = f"python3 train.py --run_number {run_number} --task_name {task_name} --layer1 {layer1} --layer2 {layer2} --embed_dim {embed_dim} --window {window} --num_heads {num_heads} --state_dim {state_dim} --save True"
    cmd += " --p 0.1 --ood_eval True --eval_p 0.5 --force_learn True"
    return cmd

for run_number in range(11):
    for experiment in [experiment2_params]:
        for layer1 in experiment["layer1"]:
            for layer2 in experiment["layer2"]:
                for embed_dim in experiment["d"]:
                    for window in experiment["window"]:
                        for num_heads in experiment["nh"]:
                            for state_dim in experiment["sd"]:
                                commands.append(make_command(layer1, layer2, embed_dim, window, num_heads, state_dim, run_number))
```

```python
def run_command(cmd):
    """Run a single shell command and return (cmd, returncode, stdout, stderr)."""
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    return cmd, result.returncode, result.stdout, result.stderr

max_workers = 20

with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = {executor.submit(run_command, cmd): cmd for cmd in commands}

    for future in as_completed(futures):
        cmd, returncode, stdout, stderr = future.result()
        print(f"[{cmd}] exited with {returncode}")
        if returncode == 1:
            executor.shutdown()
            print(stdout)
            print(stderr)
            assert False
```
