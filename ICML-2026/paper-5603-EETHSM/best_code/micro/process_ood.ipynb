```python
import numpy as np
import torch
import matplotlib.pyplot as plt

from plt_utils import plot, savefig

import os

task_name, data_name = "var_copy", "data_100_5_30"
```

```python
dashed_task_name = '-'.join(task_name.split('_'))

all_losses = {}
all_accs = {}
all_params = {}
all_eval_losses = {}
all_eval_accs = {}
```

```python
for run_name in os.listdir("results_ood/" + task_name + "/" + data_name):
    if len(run_name.split("_")) != 10: continue

    for run in os.listdir("results_ood/" + task_name + "/" + data_name + "/" + run_name):
        if run.startswith("."):
            continue
        
        # print("results_ood/" + task_name + "/" + data_name + "/" + run_name + "/" + run)
        data = torch.load("results_ood/" + task_name + "/" + data_name + "/" + run_name + "/" + run, weights_only=False)
        losses = data["losses"]
        accs = data["accs"]
        # args = data["args"]
        params = data["param_count"]
        eval_loss = data["eval_loss"]
        eval_acc = data["eval_acc"].item()

        if run_name not in all_losses.keys():
            all_losses[run_name] = losses.unsqueeze(0)
            all_eval_losses[run_name] = [eval_loss]
            all_accs[run_name] = accs.unsqueeze(0)
            all_eval_accs[run_name] = [eval_acc]
            all_params[run_name] = params
        else:
            all_accs[run_name] = torch.cat((all_accs[run_name], accs.unsqueeze(0)))
            all_eval_accs[run_name].append(eval_acc)
            all_eval_losses[run_name].append(eval_loss)
            all_losses[run_name] = torch.cat((all_losses[run_name], losses.unsqueeze(0)))

for run_name in all_losses.keys():
    all_losses[run_name] = all_losses[run_name].detach().numpy()
    all_accs[run_name] = all_accs[run_name].detach().numpy()
    all_eval_losses[run_name] = np.array(all_eval_losses[run_name])
    all_eval_accs[run_name] = np.array(all_eval_accs[run_name])
```

```python
for run_name in all_losses.keys():
    layer1 = run_name.split("_")[2]
    layer2 = run_name.split("_")[3]
    # if layer1 == "SSM" and layer2 == "SSM":
    print(run_name, np.mean(all_accs[run_name][:,-1]), np.mean(all_eval_accs[run_name]))
```
