```python
from sklearn.datasets import load_breast_cancer
from praxis import PRAXIS, ThresholdGuessBinarizer

LAMBDA_REG = 0.005
DEPTH_BUDGET = 7
RASHOMON_MULT = 0.03
LOOKAHEAD_K = 1

data = load_breast_cancer()
X_raw = data.data
y = data.target

feature_descriptions = {
    name: f"Description for {name} can be added here."
    for name in data.feature_names
}

feature_descriptions.update({
    "mean radius": "Mean distance from the center to points on the perimeter.",
    "mean texture": "Standard deviation of grayscale values.",
    "mean perimeter": "Mean size of the tumor perimeter.",
    "mean area": "Mean tumor area.",
    "mean smoothness": "Mean local variation in radius lengths.",
    "mean compactness": "Mean compactness, computed from perimeter and area.",
    "mean concavity": "Mean severity of concave portions of the contour.",
    "mean concave points": "Mean number of concave portions of the contour.",
    "mean symmetry": "Mean symmetry of the cell nucleus.",
    "mean fractal dimension": "Mean fractal dimension of the cell nucleus boundary.",
})

binarizer = ThresholdGuessBinarizer(
    n_estimators=150,
    max_depth=2,
    random_state=0,
    column_elimination=False,
)

X_bin = binarizer.fit_transform(X_raw, y)

model = PRAXIS()
model.fit(
    X_bin,
    y,
    lambda_reg=LAMBDA_REG,
    depth_budget=DEPTH_BUDGET,
    rashomon_mult=RASHOMON_MULT,
    lookahead_k=LOOKAHEAD_K,
    key_mode="hash",
)

print("min objective:", model.get_min_objective())
print("num trees:", model.count_trees())

model.save_builder_html(
    "breast_cancer_rashomon_builder.html",
    binarizer=binarizer,
    raw_feature_names=data.feature_names,
    lambda_reg=LAMBDA_REG,
    depth_budget=DEPTH_BUDGET,
    rashomon_mult=RASHOMON_MULT,
    lookahead_k=LOOKAHEAD_K,
    root_n=X_bin.shape[0],
    feature_descriptions=feature_descriptions,
)

print("Saved breast_cancer_rashomon_builder.html")
print("Open the HTML file directly in your browser.")

model.save_builder_payload(
    "breast_cancer_praxis_payload.json",
    binarizer=binarizer,
    raw_feature_names=data.feature_names,
    lambda_reg=LAMBDA_REG,
    depth_budget=DEPTH_BUDGET,
    rashomon_mult=RASHOMON_MULT,
    lookahead_k=LOOKAHEAD_K,
    root_n=X_bin.shape[0],
    feature_descriptions=feature_descriptions,
)

print("Saved breast_cancer_praxis_payload.json")
print("You can also upload this JSON file at: https://interactive-tree-builder.vercel.app")
```
