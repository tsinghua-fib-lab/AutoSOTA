For the LTL task, we identify an imbalance in the training set that affects the performance in 0 AP and 1 AP scenarios.
Specifically, the training dataset is underrepresented in these cases, which affects performance of the baselines.

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['figure.dpi'] = 192
mpl.rcParams['svg.fonttype'] = 'none'  # Critical: don't convert text to paths
plt.rcParams['pdf.use14corefonts'] = True
```

```python
distribution = {
	# Mapping from atomic proposition count to sample count
	0: 1,
	1: 11635,
	2: 103126,
	3: 222747,
	4: 284238,
	5: 178162,
}

added_distribution = {
	# This distribution was added on top of the previous one.
	0: 202,
	1: 10839,
	2: 0,
	3: 0,
	4: 0,
	5: 0,
}

ap_counts = list(distribution.keys())
sample_counts = list(distribution.values())
added_counts = [added_distribution.get(k, 0) for k in ap_counts]

plt.figure(figsize=(6, 4))
plt.bar(ap_counts, sample_counts, color="#4C78A8", label="Original")
plt.bar(ap_counts, added_counts, bottom=sample_counts, color="#F58518", label="Added")
plt.xlabel("AP count")
plt.ylabel("Sample count")
plt.title("Training set distribution by AP count")
plt.xticks(ap_counts)
plt.yscale("log")
plt.legend(frameon=False)
plt.tight_layout()

plt.savefig("ltl-imbalance.pdf")
```
