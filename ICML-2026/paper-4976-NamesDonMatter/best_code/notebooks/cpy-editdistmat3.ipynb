```python
import torch
import numpy as np

from tqdm.auto import tqdm
import seaborn as sn
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

device = torch.device('cuda')
```

```python
eval_results = torch.zeros(28, 28)
minaps = 3
# AP dim first, length dim second
apticks = list(range(minaps, 30+1, 3))
lenticks = list(range(minaps, 30+1, 3))
all_predictions = {}
```

```python
mpl.rcParams['hatch.linewidth'] = 8.0
mpl.rcParams['figure.dpi'] = int(192 * 1.25)
```

```python
fig = plt.figure(figsize=(4,4))
ax = sn.heatmap(
    eval_results,
    vmin=0, vmax=1,
    cbar=False,
)
plt.yticks([0.5+x for x in list(range(0, 28, 3))], list(range(3, 31, 3)), rotation=0)
plt.xticks([0.5+x for x in list(range(0, 28, 3))], list(range(3, 31, 3)), rotation=0)
ax.set_ylabel("Unique Characters")
ax.set_xlabel("Length")
plt.tight_layout()
plt.savefig("cpy-intro-1.png")
```
