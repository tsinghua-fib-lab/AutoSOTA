# CatBoost Experiment

This notebook runs the CRDA experiments using a **CatBoost** base regressor
(Appendix I / Table 10). It reports the percent change in test MSE across the
benchmark datasets, where a reduction is better.

---

```python
import sys
import os
sys.path.append(os.path.dirname(os.getcwd()))

from src.utils.config import Config
from src.utils.logger import Logger
from src.experiment import Experiment
from time import perf_counter
```

```python
def run(config):
    logger = Logger(log_to_file=False, log_to_console=False)
    experiment = Experiment(config, logger)
    start_time = perf_counter()
    experiment.run()
    end_time = perf_counter()
    print(f"Time taken: {end_time - start_time:.2f} seconds")
    experiment.plot_results()
```
