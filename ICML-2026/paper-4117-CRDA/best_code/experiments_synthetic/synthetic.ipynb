# Synthetic Sample-Size-Scaling Experiment

This notebook produces **Figure 3**: a controlled study on synthetic data with a
known additive-noise data-generating process

$$Y = X_1^2 + X_2 X_3 + Z, \qquad Z \perp (X_1, X_2, X_3),$$

generated by `data/anm_gen.py` (50,000 samples; see `../data/anm_data.csv`). CRDA
is applied across a range of training-set sizes for both the MLP and XGBoost base
models to reveal the "sweet spot" where it delivers the largest test-MSE reduction.

---

```python
import sys
import os
sys.path.append(os.path.dirname(os.getcwd()))

from src.utils.config import Config
from src.utils.logger import Logger
from src.experiment import Experiment


def run(config, logger):
    experiment = Experiment(config, logger)
    experiment.run()
    experiment.plot_results()
```

```python
logger = Logger(log_to_file=True, log_to_console=False, log_file="../experiments_synthetic/anm/run0.log")

run(Config(
    baseline="xgboost",
    dataset_path="../data/anm_data.csv",
    results_dir="../experiments_synthetic/anm",
    sample_sizes=[100, 250, 500, 1000, 2500],
    num_seeds=0,
    hyperparam_tune=False,
    method_param_tune=True,
    ignore_filter=True,
    save_plots=False
), logger)
```
