# Device

```python
from core.utils.utils import get_device
```

```python
DEVICE = get_device()
DEVICE
```

# DAVE Explainer

```python
from pathlib import Path
from core.explainer import DAVEExplainer
```

```python
MODEL_CONFIG: Path = Path('models_configs')/'deit3_b16_224.yaml'
```

```python
dave_explainer = DAVEExplainer(
    model_cfg_path=MODEL_CONFIG,
    device=DEVICE,
)
```

# Data Loader

```python
from torch.utils.data import DataLoader
from core.utils.data import ImageNetToySampleDataset
```

```python
DATA_DIR:   Path = Path('data_samples/imagenet_sample')
BATCH_SIZE: int  = 10
```

```python
transform = dave_explainer.input_transform

dataset = ImageNetToySampleDataset(
    DATA_DIR,
    transform=transform,
)
```

```python
data_loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
)
```

# Attribute and visualize

```python
from tqdm import tqdm
from core.utils.visualization import visualize_attribution_batch
```

```python
NUM_STEPS: int = 100
```

```python
for x, y in tqdm(
    data_loader,
    desc="Attributing...",
    total=len(data_loader),
):
    x, y = x.to(DEVICE), y.to(DEVICE)
    
    a = dave_explainer.explain(
        x=x, 
        y=y, 
        num_steps=NUM_STEPS,
        post_proc=True,
    )

    visualize_attribution_batch(
        attribution_map=a, 
        image_batch=x, 
        input_transform=transform,
    )
```
